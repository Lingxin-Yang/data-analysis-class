#### Python 正则表达式练习

```python
import re

# 1. 检查字符串是否仅包含英文字母a-z、A-Z和数字0-9
def check_alnum(s):
    pattern = r'^[a-zA-Z0-9]+$'
    return bool(re.fullmatch(pattern, s))

# 2. 匹配一个a后面跟着零个或多个b
def match_a_zero_more_b(s):
    pattern = r'^ab*$'
    return bool(re.fullmatch(pattern, s))

# 3. 匹配一个a后面跟着一个或多个b
def match_a_one_more_b(s):
    pattern = r'^ab+$'
    return bool(re.fullmatch(pattern, s))

# 4. 匹配以a开头、以b结尾，中间任意字符
def match_start_a_end_b(s):
    pattern = r'^a.*b$'
    return bool(re.fullmatch(pattern, s))

# 5. 删除IP地址各段前导零，全0保留一个0
def ip_remove_leading_zero(ip_str):
    def sub_zero(match):
        seg = match.group(1)
        return str(int(seg))
    return re.sub(r'\b0*(\d+)\b', sub_zero, ip_str)

# 6. 空格换下划线，下划线换空格（交换）
def swap_space_underline(s):
    s1 = re.sub(r' ', '\x01', s)
    s2 = re.sub(r'_', ' ', s1)
    s3 = re.sub(r'\x01', '_', s2)
    return s3

# 7. yyyy-mm-dd 转换 dd-mm-yyyy
def date_ymd_to_dmy(date_str):
    return re.sub(r'^(\d{4})-(\d{2})-(\d{2})$', r'\3-\2-\1', date_str)

# 8. 找出所有以a或e开头的单词
def find_word_start_ae(s):
    return re.findall(r'\b[aeAE]\w*\b', s)

# 9. Road缩写为Rd
def road_to_rd(s):
    return re.sub(r'\bRoad\b', 'Rd.', s)

# 10. 找出长度≥4的单词
def find_word_len4(s):
    return re.findall(r'\b\w{4,}\b', s)

# 11. 删除除字母数字以外所有字符
def keep_alnum_only(s):
    return re.sub(r'[^a-zA-Z0-9]', '', s)

# 12. 删除括号以及括号里面全部内容
def remove_bracket_content(s):
    return re.sub(r'\(.*?\)', '', s).strip()



if __name__ == "__main__":
    print("任务1")
    s1_1 = "abc123"
    s1_2 = "abc-123"
    print(f'输入"{s1_1}"，结果：{check_alnum(s1_1)}')
    print(f'输入"{s1_2}"，结果：{check_alnum(s1_2)}')

    print("\n任务2")
    s2_1 = "a"
    s2_2 = "abb"
    s2_3 = "acb"
    print(f'输入"{s2_1}"，结果：{match_a_zero_more_b(s2_1)}')
    print(f'输入"{s2_2}"，结果：{match_a_zero_more_b(s2_2)}')
    print(f'输入"{s2_3}"，结果：{match_a_zero_more_b(s2_3)}')

    print("\n任务3")
    s3_1 = "a"
    s3_2 = "abb"
    s3_3 = "ab"
    print(f'输入"{s3_1}"，结果：{match_a_one_more_b(s3_1)}')
    print(f'输入"{s3_2}"，结果：{match_a_one_more_b(s3_2)}')
    print(f'输入"{s3_3}"，结果：{match_a_one_more_b(s3_3)}')

    print("\n任务4")
    s4_1 = "a123b"
    s4_2 = "a123c"
    s4_3 = "x123b"
    print(f'输入"{s4_1}"，结果：{match_start_a_end_b(s4_1)}')
    print(f'输入"{s4_2}"，结果：{match_start_a_end_b(s4_2)}')
    print(f'输入"{s4_3}"，结果：{match_start_a_end_b(s4_3)}')

    print("\n任务5")
    s5 = "192.168.001.010"
    print(f'输入"{s5}"，输出：{ip_remove_leading_zero(s5)}')

    print("\n任务6")
    s6 = "hello world_test"
    print(f'输入"{s6}"，输出：{swap_space_underline(s6)}')

    print("\n任务7")
    s7 = "2025-12-25"
    print(f'输入"{s7}"，输出：{date_ymd_to_dmy(s7)}')

    print("\n任务8")
    s8 = "apple egg cat End ant"
    print(f'输入"{s8}"，匹配结果：{find_word_start_ae(s8)}')

    print("\n任务9")
    s9 = "Green Road, Main Road"
    print(f'输入"{s9}"，输出：{road_to_rd(s9)}')

    print("\n任务10")
    s10 = "hi hello test a cat"
    print(f'输入"{s10}"，匹配结果：{find_word_len4(s10)}')

    print("\n任务11")
    s11 = "abc@123!xyz"
    print(f'输入"{s11}"，输出：{keep_alnum_only(s11)}')

    print("\n任务12")
    test_list = ["example (.com)", "w3resource", "github (.com)", "stackoverflow (.com)"]
    for item in test_list:
        print(f'输入"{item}"，输出："{remove_bracket_content(item)}"')
```

#### 从⽹⻚检索⽓候数据

```python
import requests
import re
import pandas as pd
from bs4 import BeautifulSoup

city_urls = {
    "Madison": "https://en.wikipedia.org/wiki/Madison,_Wisconsin",
    "Buenos Aires": "https://en.wikipedia.org/wiki/Buenos_Aires",
    "Harbin": "https://en.wikipedia.org/wiki/Harbin"
}

headers = {
    "User-Agent": (
        "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
        "AppleWebKit/537.36 Chrome/120.0 Safari/537.36"
    )
}

soup_cache = {}
table_cache = {}

def get_soup(url):

    if url in soup_cache:
        return soup_cache[url]

    try:
        response = requests.get(
            url,
            headers=headers,
            timeout=20
        )

        if response.status_code != 200:
            print("网页访问失败，状态码：", response.status_code)
            soup_cache[url] = None
            return None

        if not response.text.strip():
            print("网页内容为空！")
            soup_cache[url] = None
            return None

        soup = BeautifulSoup(response.text, "html.parser")

        if soup.find("html") is None:
            print("HTML 解析失败！")
            soup_cache[url] = None
            return None

        soup_cache[url] = soup

        return soup

    except requests.exceptions.Timeout:
        print("网页访问超时！")
        soup_cache[url] = None
        return None

    except requests.exceptions.RequestException as e:
        print("网页请求异常：", e)
        soup_cache[url] = None
        return None

def get_direct_children(tag):
    return tag.find_all(recursive=False)

def get_table_rows(table):
    return table.find_all("tr")

def get_row_cells(row):
    return row.find_all(["th", "td"], recursive=False)

def get_cell_text(cell):
    return cell.get_text(" ", strip=True).strip()

def is_title_row(row):

    cells = get_row_cells(row)

    if not cells:
        return False

    text = " ".join(get_cell_text(c) for c in cells)

    patterns = [
        r"Climate\s+data\s+for",
        r"Parámetros\s+climáticos",
        r"Tabla\s+climática",
        r"Climatic\s+data\s+for"
    ]

    return any(
        re.search(pattern, text, re.I)
        for pattern in patterns
    )

def is_month_header(row):

    cells = get_row_cells(row)

    if not cells:
        return False

    texts = [
        get_cell_text(cell).lower().rstrip(".")
        for cell in cells
    ]

    english_months = {
        "jan", "january",
        "feb", "february",
        "mar", "march",
        "apr", "april",
        "may",
        "jun", "june",
        "jul", "july",
        "aug", "august",
        "sep", "september",
        "oct", "october",
        "nov", "november",
        "dec", "december"
    }

    spanish_months = {
        "ene", "enero",
        "febrero",
        "mar", "marzo",
        "abr", "abril",
        "mayo",
        "jun", "junio",
        "jul", "julio",
        "ago", "agosto",
        "sep", "septiembre",
        "oct", "octubre",
        "nov", "noviembre",
        "dic", "diciembre"
    }

    all_months = english_months | spanish_months

    month_count = sum(
        1 for text in texts
        if text in all_months
    )

    first_text = texts[0]

    if first_text in {"month", "mes"}:
        return True

    if month_count >= 3:
        return True

    return False

def is_source_row(row):

    cells = get_row_cells(row)

    if not cells:
        return False

    text = " ".join(
        get_cell_text(cell) for cell in cells
    ).lower()

    return bool(
        re.match(
            r"^\s*(source|sources|fuente|fuentes)\s*:",
            text
        )
    )

def is_climate_table(table):

    rows = get_table_rows(table)

    if len(rows) < 3:
        return False

    table_text = table.get_text(" ", strip=True)

    english_title = bool(
        re.search(
            r"Climate\s+data\s+for",
            table_text,
            re.I
        )
    )

    other_title = bool(
        re.search(
            r"Parámetros\s+climáticos|"
            r"Tabla\s+climática|"
            r"Climatic\s+data\s+for",
            table_text,
            re.I
        )
    )

    month_exists = any(
        is_month_header(row)
        for row in rows
    )

    metric_pattern = re.compile(
        r"temperature|temperatura|"
        r"precipitation|precipitación|"
        r"humidity|humedad|"
        r"snowfall|snow|nieve|"
        r"sunshine|sun|sol|"
        r"record high|record low|"
        r"mean daily|daily mean|"
        r"temperatura media|"
        r"lluvia|precipitaciones",
        re.I
    )

    metric_count = 0

    for row in rows:

        cells = get_row_cells(row)

        if not cells:
            continue

        first_text = get_cell_text(cells[0])

        if metric_pattern.search(first_text):
            metric_count += 1

    if month_exists and (english_title or other_title):
        return True

    if month_exists and metric_count >= 3:
        return True

    return False

def inspect_table_structure(table, city):
    print("城市：", city)

    children = get_direct_children(table)

    print("直接子节点数量：", len(children))

    print("\n直接子节点标签名称：")

    for index, child in enumerate(children, start=1):
        print(f"{index}. <{child.name}>")

    tag_counts = {}

    for child in children:
        tag_counts[child.name] = (
            tag_counts.get(child.name, 0) + 1
        )

    print("\n直接子节点标签统计：")

    for tag, count in tag_counts.items():
        print(f"<{tag}>：{count} 个")

    print("\n表格 class 属性：")
    print(table.get("class", []))

    rows = get_table_rows(table)

    print("\n表格内部 tr 数量：", len(rows))

    if rows:

        first_row = rows[0]

        first_children = get_direct_children(first_row)

        print("\n第一行结构：")

        print("第一行直接子节点数量：", len(first_children))

        print(
            "第一行直接子节点标签：",
            [child.name for child in first_children]
        )

        print(
            "第一行单元格文本：",
            [get_cell_text(cell)
             for cell in get_row_cells(first_row)]
        )

# 任务一：网页访问、HTML 解析及表格节点结构观察

print("任务一\n\n\n")

for city, url in city_urls.items():

    print("\n" + "-" * 65)
    print("当前城市：", city)
    print("网页地址：", url)

    soup = get_soup(url)

    if soup is None:
        print("检测失败：网页访问或 HTML 解析失败！")
        continue

    print("检测成功：网页访问成功！")
    print("检测成功：HTML 解析成功！")

    if soup.title:
        print("网页标题：", soup.title.get_text(strip=True))

    all_tables = soup.find_all("table")

    print("\n网页中的表格数量：", len(all_tables))

    if not all_tables:
        print("检测失败：网页中没有表格！")
        continue

    print("任务一网页检测：PASS")
    inspect_table_structure(all_tables[0], city)

# 任务二：编写 retrieve_climate_table(s)
print("\n\n\n任务二\n" )

def retrieve_climate_table(s):

    if not isinstance(s, str) or not s.strip():
        print("检测失败：URL 无效！")
        return None

    if s in table_cache:
        return table_cache[s]

    soup = get_soup(s)

    if soup is None:
        print("检测失败：无法获取网页！")
        table_cache[s] = None
        return None

    tables = soup.find_all("table")

    print("网页中检测到表格数量：", len(tables))

    for table in tables:

        if is_climate_table(table):

            rows = get_table_rows(table)

            print("检测成功：找到气候数据表格！")
            print("表格行数：", len(rows))
            print("表格标题：")

            caption = table.find("caption")

            if caption:
                print(get_cell_text(caption))
            else:
                print("通过表格内部标题和气候数据结构识别")

            table_cache[s] = table

            return table

    print("检测失败：没有找到气候数据表格！")

    table_cache[s] = None

    return None

def climate_table_to_dataframe(table):

    matrix = []

    for row in get_table_rows(table):

        if is_title_row(row):
            continue

        cells = get_row_cells(row)

        if not cells:
            continue

        values = [
            get_cell_text(cell)
            for cell in cells
        ]

        if any(values):
            matrix.append(values)

    if not matrix:
        return None

    max_columns = max(
        len(row) for row in matrix
    )

    for row in matrix:
        row.extend(
            [""] * (max_columns - len(row))
        )

    if is_month_header_from_values(matrix[0]):

        columns = matrix[0]
        data = matrix[1:]

    else:

        columns = [
            "气候指标"
        ] + [
            f"数据列{i}"
            for i in range(2, max_columns + 1)
        ]

        data = matrix

    df = pd.DataFrame(
        data,
        columns=columns
    )

    return df

def is_month_header_from_values(values):

    fake_row = BeautifulSoup(
        "<tr>" + "".join(
            f"<th>{value}</th>"
            for value in values
        ) + "</tr>",
        "html.parser"
    ).tr

    return is_month_header(fake_row)

task2_results = {}
table_signatures = {}

for city, url in city_urls.items():

    
    print("当前测试城市：", city)
    

    table = retrieve_climate_table(url)

    if table is None:

        task2_results[city] = False

        print("任务二检测结果：FAIL")

        continue

    task2_results[city] = True

    inspect_table_structure(table, city)

    children = get_direct_children(table)

    signature = tuple(
        child.name for child in children
    )

    table_signatures[city] = signature

    df = climate_table_to_dataframe(table)

    if df is not None:

        print("\n完整气候数据表格：\n")

        with pd.option_context(
            "display.max_columns", None,
            "display.max_rows", None,
            "display.width", 200,
            "display.max_colwidth", 30
        ):

            print(df.to_string(index=False))

        print("\n表格数据行数：", len(df))
        print("表格数据列数：", len(df.columns))

        print("任务二检测结果：PASS")

    else:

        task2_results[city] = False

        print("检测失败：DataFrame 转换失败！")
        print("任务二检测结果：FAIL")

print("三个城市气候表格 HTML 结构比较")

for city, signature in table_signatures.items():

    print("\n城市：", city)
    print("直接子节点结构：", signature)

if len(table_signatures) >= 2:

    unique_signatures = set(
        table_signatures.values()
    )

    if len(unique_signatures) > 1:

        print("\n检测结果：存在不同的 HTML 表格结构！")
        print("说明程序测试了不同结构的气候数据表格。")

    else:

        print("\n检测结果：本次获取的表格直接子节点结构相同。")
        print("请结合上面的第一行结构、class 属性及实际 HTML 继续比较。")

else:

    print("\n检测结果：成功提取的城市数量不足，无法完成结构比较。")

# 任务三：编写 list_climate_table_row_names(s)
print("\n\n\n任务三\n" )
def list_climate_table_row_names(s):

    if not isinstance(s, str) or not s.strip():
        return None

    table = retrieve_climate_table(s)

    if table is None:
        return None

    row_names = []

    for row in get_table_rows(table):

        cells = get_row_cells(row)

        if not cells:
            continue

        if is_title_row(row):
            continue

        if is_month_header(row):
            continue

        if is_source_row(row):
            continue

        row_header = None

        for cell in cells:

            if cell.name != "th":
                continue

            if cell.get("scope", "").lower() == "row":

                row_header = cell
                break

        if row_header is None:

            for cell in cells:

                if cell.name == "th":

                    row_header = cell
                    break

        if row_header is None:
            continue

        name = get_cell_text(row_header)

        if not name:
            continue

        if re.search(
            r"Climate\s+data\s+for|"
            r"Parámetros\s+climáticos|"
            r"Tabla\s+climática|"
            r"^Source\s*:|"
            r"^Sources\s*:|"
            r"^Fuente\s*:|"
            r"^Fuentes\s*:",
            name,
            re.I
        ):
            continue

        name = str(name).strip()

        if name:
            row_names.append(name)

    if not row_names:
        return None

    return row_names

print("三个城市气候数据表格行名称提取")

task3_results = {}

for city, url in city_urls.items():


    print("当前处理城市：", city)
 

    names = list_climate_table_row_names(url)

    if names is not None:

        task3_results[city] = True

        print("检测成功：行名称提取成功！")
        print("行名称数量：", len(names))

        print("\n提取结果：")

        for index, name in enumerate(names, start=1):
            print(f"{index}. {name}")

        assert isinstance(names, list)

        assert all(
            isinstance(name, str)
            for name in names
        )

        assert all(
            name == name.strip()
            for name in names
        )

        assert not any(
            re.search(
                r"Climate\s+data\s+for|"
                r"Parámetros\s+climáticos|"
                r"Tabla\s+climática",
                name,
                re.I
            )
            for name in names
        )

        assert not any(
            re.match(
                r"^(Source|Sources|Fuente|Fuentes)\s*:",
                name,
                re.I
            )
            for name in names
        )

        print("\n任务三检测结果：PASS")

    else:

        task3_results[city] = False

        print("检测结果：没有获取到行名称！")
        print("任务三检测结果：FAIL")
```
