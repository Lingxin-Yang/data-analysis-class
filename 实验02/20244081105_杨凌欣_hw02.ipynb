# 字符串的乐趣

```python
# 回文数

print("任务一")
def is_palindrome(s: str) -> bool:
    cleaned_s = s.replace(" ", "").lower()
    # 检查清理后的字符串是否等于其反转结果
    return cleaned_s == cleaned_s[::-1]

# 检验函数
def test_is_palindrome():
    test_inputs = [
        "tacocat",
        " T A C O cat ",
        "Racecar",
        "123 21",
        "hello world",
        "12a 321",
        "a",
        "159951753357",
        "ada000dgoogd000ada"
    ]
    
    for s in test_inputs:
        print(is_palindrome(s))

# 运行检验代码
test_is_palindrome()



# 任务二：字母是否是顺序的
print("\n任务二")
def is_abecedarian(word):
    word = word.replace(" ", "")  # 去掉所有空格
    for i in range(len(word) - 1):
        if word[i] > word[i + 1]:
            return False

    return True


# 检验代码
print(is_abecedarian("adder"))       
print(is_abecedarian("beet"))       
print(is_abecedarian("dog"))         
print(is_abecedarian("cat"))         
print(is_abecedarian("abcd efgh xyz"))  
print(is_abecedarian("hello world")) 
print(is_abecedarian(""))            



#任务三：去掉元音字母
print("\n任务三")
def remove_vowels(word):
    result = ""

    for ch in word:
        if ch not in "aeiou":
            result += ch

    return result


# 检验代码
print(remove_vowels("cat"))       
print(remove_vowels("audio"))     
print(remove_vowels("hello"))     
print(remove_vowels("python"))    
print(remove_vowels("aeiou"))     
print(remove_vowels("aaagh"))          
```

# 列表的乐趣

```python
# 任务一：列表反转
print("\n任务四")
def list_reverse(lst):
    if not isinstance(lst, list):
        raise TypeError("输入必须是一个列表")

    result = []

    for i in range(len(lst) - 1, -1, -1):
        result.append(lst[i])

    return result


# 检验代码
print(list_reverse([1, 2, 3]))      
print(list_reverse(["a", "b", "c"])) 
print(list_reverse([10]))          
print(list_reverse([159951]))              

# 检验输入不是列表的情况
try:
    print(list_reverse("abc"))
except TypeError as e:
    print(e)



# 任务二：检查序列是否有序
print("\n任务二")


def is_sorted(seq):
    for i in range(len(seq) - 1):
        if seq[i] > seq[i + 1]:
            return False

    return True


# 检验代码
print(is_sorted([1, 2, 3, 4, 5]))  # True
print(is_sorted([1, 2, 2, 3, 5]))  # True
print(is_sorted([1, 3, 2, 4, 5]))  # False
print(is_sorted([5, 4, 3, 2, 1]))  # False
print(is_sorted([]))               # True
print(is_sorted([10]))             # True
print(is_sorted("abcdef"))         # True
print(is_sorted("abdc"))           # False
```

# 作为向量的元组

```python
# 1. 向量标量乘法
def vec_scalar_mult(t, s):
    if not isinstance(t, tuple):
        raise TypeError("第一个参数必须是元组")
    if not isinstance(s, (int, float)):
        raise TypeError("标量s必须是整数或浮点数")
    new_list = []
    for elem in t:
        if not isinstance(elem, (int, float)):
            raise TypeError("元组内元素只能是整数或浮点数")
        new_list.append(elem * s)
    return tuple(new_list)


# 2. 向量内积
def vec_inner_product(x, y):
    if not isinstance(x, tuple) or not isinstance(y, tuple):
        raise TypeError("两个输入必须都是元组")
    if len(x) != len(y):
        raise ValueError("两个向量长度不一致")
    total = 0.0
    for xi, yi in zip(x, y):
        if not isinstance(xi, (int, float)) or not isinstance(yi, (int, float)):
            raise TypeError("向量元素只能是整数或浮点数")
        total += xi * yi
    return total


# 3. 校验有效矩阵
def check_valid_mx(mx):
    if not isinstance(mx, tuple):
        return False
    if len(mx) == 0:
        return True
    row_len = len(mx[0])
    for row in mx:
        if not isinstance(row, tuple):
            return False
        if len(row) != row_len:
            return False
        for elem in row:
            if not isinstance(elem, (int, float)):
                return False
    return True


# 4. 矩阵乘向量
def mx_vec_mult(mx, vec):
    if not check_valid_mx(mx):
        raise TypeError("输入不是有效的矩阵")
    if not isinstance(vec, tuple):
        raise TypeError("向量必须是元组")
    if len(mx) == 0:
        return ()
    n_cols = len(mx[0])
    if len(vec) != n_cols:
        raise ValueError("矩阵的列数必须等于向量的维度")
    result = []
    for row in mx:
        result.append(vec_inner_product(row, vec))
    return tuple(result)


if __name__ == "__main__":
    print("任务一")
    print(vec_scalar_mult((1,2,3),2))
    print(vec_scalar_mult((1,2,3),1.5))

    print("\n任务二")
    print(vec_inner_product((1,2,3),(4,5,6)))

    print("\n任务三")
    mat1 = ((1, 2), (3, 4))
    mat2 = ((1, 2), (3, 4, 5))
    mat3 = [(1, 2), (3, 4)]
    print(check_valid_mx(((1,2),(3,4))))
    print(check_valid_mx(((1,2),(3,4,5))))
    print(check_valid_mx([(1,2),(3,4)]))

    print("\n任务四")
    mat = ((1, 2), (3, 4))
    vec = (5, 6)
    print(mx_vec_mult(((1,2),(3,4)),(5,6)))

```

# 更多向量乐趣

```python
class Vector:
    def __init__(self, dim, elements=None):
        if not isinstance(dim, int):
            raise TypeError("维度必须是整数")
        if dim < 0:
            raise ValueError("维度不能为负数")


        if elements is None:
            elements = [0] * dim
        else:
            elements = list(elements)
            for e in elements:
                if not isinstance(e, (int, float)):
                    raise TypeError("元素只能是整数或浮点数")

            if len(elements) != dim:
                raise ValueError("维度与元素个数不一致")

        self.dim = dim
        self.elements = elements

    # 比较运算符 ==, <, <=, >, >=
    def __eq__(self, other):
        if not isinstance(other, Vector):
            raise TypeError("只能和Vector对象比较")
        if self.dim != other.dim:
            raise ValueError("维度不同，无法比较")
        return self.elements == other.elements

    def __lt__(self, other):
        if not isinstance(other, Vector):
            raise TypeError("只能和Vector对象比较")
        if self.dim != other.dim:
            raise ValueError("维度不同，无法比较")
        return self.elements < other.elements

    def __le__(self, other):
        return self < other or self == other

    def __gt__(self, other):
        return not (self <= other)

    def __ge__(self, other):
        return not (self < other)

    # 任务4：dot内积方法
    def dot(self, other):
        if not isinstance(other, Vector):
            raise TypeError("参数必须是Vector对象")
        if self.dim != other.dim:
            raise ValueError("两个向量维度不一致，不能计算内积")
        total = 0.0
        for a, b in zip(self.elements, other.elements):
            total += a * b
        return total

    # 任务5：* 乘法运算符
    def __mul__(self, other):
        if isinstance(other, (int, float)):
            new_elems = [e * other for e in self.elements]
            return Vector(self.dim, new_elems)
        elif isinstance(other, Vector):
            if self.dim != other.dim:
                raise ValueError("两个向量维度不一致，不能逐元素相乘")
            new_elems = [a * b for a, b in zip(self.elements, other.elements)]
            return Vector(self.dim, new_elems)
        else:
            raise TypeError("乘法仅支持标量或Vector对象")

    def __rmul__(self, other):
        return self * other


if __name__ == "__main__":
    print("\n任务1")
    v1 = Vector(2, (2, 4))
    print(v1.elements)

    print("\n任务2")
    print("维度和元素应是实例属性。因为每个向量是独立对象，拥有各自的维度与元素；若设为类属性，所有向量会共享数据，互相干扰。")
    print("\n任务3")
    v2 = Vector(2, (2, 5))
    print(v1 < v2)
    print(v1 == v2)

    print("\n任务4")
    v3 = Vector(3, (1,2,3))
    v4 = Vector(3, (4,5,6))
    print(v3.dot(v4))

    print("\n任务5")
    v5 = Vector(2, (2,3))
    v6 = Vector(2, (4,5))
    print((v5 * v6).elements)
    print((2 * v5).elements)
```
