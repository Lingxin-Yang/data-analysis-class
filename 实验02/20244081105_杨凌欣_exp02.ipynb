## 实验 1：NumPy 数组、数据类型与索引


```python
print("任务一")
int_arr = np.array([1, 2, 3, 4, 5])
float_arr = np.array([1.5, 2.5, 3.5, 4.5])
bool_arr = np.array([True, False, True, True])

for name, arr in [("整数数组", int_arr), ("浮点数数组", float_arr), ("布尔数组", bool_arr)]:
    print(f"{name}: {arr}")
    print(f"  dtype = {arr.dtype}, shape = {arr.shape}, ndim = {arr.ndim}")



print("任务二")
arr12 = np.arange(12)
arr_3x4 = arr12.reshape(3, 4)
arr_2x3x2 = arr12.reshape(2, 3, 2)

print("原一维数组:")
print(arr12, "shape =", arr12.shape, "size =", arr12.size)
print("\nreshape(3, 4):")
print(arr_3x4)
print("shape =", arr_3x4.shape, "ndim =", arr_3x4.ndim)
print("\nreshape(2, 3, 2):")
print(arr_2x3x2)
print("shape =", arr_2x3x2.shape, "ndim =", arr_2x3x2.ndim)

try:
    arr12.reshape(3, 5)
except ValueError as e:
    print("\nreshape(3, 5) 失败，原因:", e)




print("任务三")
mat = np.arange(12).reshape(3, 4)
print("原始 (3, 4) 数组:")
print(mat)
row_2 = mat[1]
print("\n第 2 行:")
print(row_2, "shape =", row_2.shape)
cols_1_to_3 = mat[:, 0:3]
print("\n第 1~3 列:")
print(cols_1_to_3, "shape =", cols_1_to_3.shape)
greater_than_6 = mat[mat > 6]
print("\n大于 6 的元素:")
print(greater_than_6, "shape =", greater_than_6.shape)



print("任务四")
arr10 = np.arange(10)
print("原始数组 arr10 =", arr10)

mask_lt2_or_gt7 = (arr10 < 2) | (arr10 > 7)
mask_gt3_and_lt7 = (arr10 > 3) & (arr10 < 7)

print("\n小于 2 或大于 7 的元素:", arr10[mask_lt2_or_gt7])
print("大于 3 且小于 7 的元素:", arr10[mask_gt3_and_lt7])


mask_or_func = np.logical_or(arr10 < 2, arr10 > 7)
mask_and_func = np.logical_and(arr10 > 3, arr10 < 7)
print("\n用 np.logical_or 筛选结果：", arr10[mask_or_func])
print("\n用 np.logical_and 筛选结果：", arr10[mask_and_func])






print("任务五")
A = np.array([[1, 2],[3, 4]])
B = np.array([[5, 6],[7, 8]])

print("A =")
print(A)
print("\nB =")
print(B)

print("\n逐元素乘法")
print("A * B =")
print(A * B)
print("np.multiply(A, B) =")
print(np.multiply(A, B))
print("两种写法结果相同:", np.array_equal(A * B, np.multiply(A, B)))

print("\n矩阵乘法")
print("A @ B =")
print(A @ B)
print("np.matmul(A, B) =")
print(np.matmul(A, B))
print("两种写法结果相同:", np.array_equal(A @ B, np.matmul(A, B)))

print("\n逐元素结果与矩阵乘法结果是否相同:", np.array_equal(A * B, A @ B))



```


## 实验 2：随机抽样、统计与缺失值


```python
import numpy as np
import pandas as pd


np.random.seed(42)

print("任务1")
scores = np.random.normal(loc=70, scale=10, size=1000)
scores = np.clip(scores, a_min=0, a_max=100)

n_sample = len(scores)
min_score = np.min(scores)
max_score = np.max(scores)
mean_score = np.mean(scores)
std_score = np.std(scores)
range_score = max_score - min_score

print(f"样本数: {n_sample}")
print(f"最小值: {min_score:.2f}")
print(f"最大值: {max_score:.2f}")
print(f"均值: {mean_score:.2f}")
print(f"标准差: {std_score:.2f}")
print(f"极差: {range_score:.2f}")





print("\n任务2")
pop = np.arange(1, 11)
print("总体数组：", pop)
sample_uniform = np.random.choice(pop, size=5, replace=False)
print("均匀不放回抽样5个：", sample_uniform)
weights = pop / np.sum(pop)
sample_weighted = np.random.choice(pop, size=5, replace=False, p=weights)
print("按数值大小加权不放回抽样5个：", sample_weighted)




print("\n任务3")
arr = np.arange(1, 11)
print("原始数组 arr：", arr)
arr_shuffled_new = np.random.permutation(arr)
print("permutation得到新数组：", arr_shuffled_new)
print("permutation后原数组不变：", arr)
arr2 = np.arange(1,11)
print("\n原地打乱前arr2：", arr2)
np.random.shuffle(arr2)
print("shuffle原地打乱后arr2：", arr2)





print("\n任务4")
scores_nan = scores.copy()
scores_nan[[9,19,29]] = np.nan
mean_ordinary = np.mean(scores_nan)         
mean_ignore_nan = np.nanmean(scores_nan)   
count_nan = np.sum(np.isnan(scores_nan))
print(f"普通np.mean均值（含nan）：{mean_ordinary}")
print(f"np.nanmean忽略缺失值均值：{mean_ignore_nan:.2f}")
print(f"缺失值个数：{count_nan}")





print("\n任务5")
df_score = pd.DataFrame({
    "index": np.arange(len(scores_nan)),
    "score": scores_nan
})
print("DataFrame前5行：")
print(df_score.head())


print("1. 成绩数组长度：", len(scores_nan) == 1000)
print("2. nan数量：", np.sum(np.isnan(scores_nan)) == 3)
print("3. DataFrame是否存在内存：", 'df_score' in locals())
print("4. 成绩范围检查（全部0~100）：", np.all((scores_nan[~np.isnan(scores_nan)] >= 0) & (scores_nan[~np.isnan(scores_nan)] <= 100)))
```


## 实验 3：SciPy 数值计算与 Matplotlib 可视化


```python
import numpy as np
from scipy import linalg, optimize, stats
import matplotlib.pyplot as plt


%matplotlib inline


print("任务1")
A = np.array([[3, 1],
              [1, 2]])
b = np.array([9, 8])
sol = linalg.solve(A, b)
x_sol, y_sol = sol[0], sol[1]
print(f"方程组解：x = {x_sol:.2f}, y = {y_sol:.2f}")
res_check = A @ sol
print(f"代入验证A@sol：{res_check}，原始b={b}")





print("\n任务2")
def f(x):
    return x**2 - 2


root_neg = optimize.root(f, x0=-2)
root_pos = optimize.root(f, x0=2)

print("负根求解：")
print(f" {root_neg.success}, 根 = {root_neg.x[0]:.4f}, f(root)={f(root_neg.x[0]):.2e}")
print("正根求解：")
print(f"{root_pos.success}, 根 = {root_pos.x[0]:.4f}, f(root)={f(root_pos.x[0]):.2e}")





print("\n任务3")
def func(x):
    return x**2 + x + 2


res_min = optimize.minimize(func, x0=0)
print(f"极小点x = {res_min.x[0]:.4f}")
print(f"极小函数值 f(x) = {res_min.fun:.4f}")
print(f"解析：极小点x=-0.5，函数值=1.75，对比数值结果")




print("\n任务4")
np.random.seed(42)
sample1 = np.random.normal(loc=0, scale=1, size=100)
sample2 = np.random.normal(loc=1, scale=1, size=100)
t_stat, p_val = stats.ttest_ind(sample1, sample2)
print(f"t统计量 = {t_stat:.4f}")
print(f"p值 = {p_val:.4f}")
alpha = 0.05
if p_val < alpha:
    print(f"p < {alpha}，拒绝原假设，两组均值存在显著差异")
else:
    print(f"p >= {alpha}，不能拒绝原假设，两组均值无显著差异")




print("\n任务5")
# 直方图
valid_scores = scores_nan[~np.isnan(scores_nan)]

plt.figure(figsize=(12,4))
plt.subplot(1,3,1)
plt.hist(valid_scores, bins=20, alpha=0.7)
plt.title("成绩直方图")
plt.xlabel("分数")
plt.ylabel("频数")

# 箱线图
plt.subplot(1,3,2)
plt.boxplot(valid_scores)
plt.title("成绩箱线图")
plt.ylabel("分数")



# 坐标系绘制
plt.subplot(1,3,3)
x_line = np.linspace(0,5,100)
y1 = x_line
y2 = x_line **2
y3 = x_line **3
plt.plot(x_line,y1,label='$y=x$')
plt.plot(x_line,y2,label='$y=x^2$')
plt.plot(x_line,y3,label='$y=x^3$')
plt.title("函数曲线")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.tight_layout()
plt.show()



print("任务1")
check1 = np.allclose(A @ sol, b)
print(f"{check1}")
print("任务2")
check2_neg = np.isclose(f(root_neg.x[0]), 0)
check2_pos = np.isclose(f(root_pos.x[0]), 0)
print(f"负根：{check2_neg}, 正根：{check2_pos}")
print("任务3")
check3_x = np.isclose(res_min.x[0], -0.5)
check3_f = np.isclose(res_min.fun, 1.75)
print(f"极小点：{check3_x}, 函数值：{check3_f}")
print("任务4")
print(f"样本1长度={len(sample1)},样本2长度={len(sample2)}，均为100：{len(sample1)==100 and len(sample2)==100}")

```


## 实验 4：pandas Series、DataFrame


```python
import pandas as pd
import numpy as np


print("任务1")
records = {
    "Name": ["Alice", "Bob", "Carol", "David", "Eva", "Frank"],
    "Group": ["A", "A", "B", "B", "A", "B"],
    "Math": [88, 72, 91, 65, 79, 84],
    "English": [90, 68, 85, 73, 92, 77],
    "Python": [95, 80, 89, 70, 86, 82],
}
df = pd.DataFrame(records)
df = df.set_index("Name")
df["Total"] = df["Math"] + df["English"] + df["Python"]
df["Average"] = df["Total"] / 3

print("行索引：", df.index.tolist())
print("列名：", df.columns.tolist())
print("\n前3行：")
print(df.head(3))
print("\n基本统计信息：")
print(df.describe())


print("\n任务2")
carol_loc = df.loc["Carol"]
print("loc获取Carol记录：")
print(carol_loc)
carol_iloc = df.iloc[2]
print("\niloc获取Carol记录：")
print(carol_iloc)


high_avg = df[df["Average"] >= 80]
print("\n平均分不低于80的学生：")
print(high_avg)




print("\n任务3")
col_mean = df[["Math","English","Python"]].mean(axis=0)
print("每科平均分：")
print(col_mean)
row_mean = df[["Math","English","Python"]].mean(axis=1)
print("\n每名学生的行均值：")
print(row_mean)




print("\n任务4")
df["MaxScore"] = df[["Math","English","Python"]].apply(lambda row: row.max(), axis=1)
print("每名学生三科最高分，新增MaxScore列：")
print(df[["Math","English","Python","MaxScore"]])
agg_result = df[["Math","English","Python"]].agg(["min", "mean", "max"])
print("\n每科统计agg结果：")
print(agg_result)
df["Math_Rank"] = df["Math"].rank(ascending=False)
print("\n数学成绩排名：")
print(df[["Math","Math_Rank"]])




print("\n任务5")
group_stat = df.groupby("Group").agg(
    人数=("Average", "count"),
    三科平均分=("Average", "mean")
)
print("按Group分组统计：")
print(group_stat)

df["Group_Python_Mean"] = df.groupby("Group")["Python"].transform("mean")
print("\n增加组内Python平均分后的前5行：")
print(df[["Group","Python","Group_Python_Mean"]].head())





print("\n任务6")
group_filtered = df.groupby("Group").filter(lambda x: x["Python"].mean() >= 80)
print("Python平均分≥80的组内全部学生：")
print(group_filtered)



print("\n任务7")
score_cols = ["Math","English","Python"]
cov_matrix = df[score_cols].cov()
corr_matrix = df[score_cols].corr()
print("协方差矩阵：")
print(cov_matrix)
print("\n皮尔森相关系数矩阵：")
print(corr_matrix)

print("\n解释Math与English")
print("协方差符号为正：两门成绩同向变化，数学高分的人英语倾向也更高；")
print("相关系数绝对值大小代表线性相关强弱，越接近1越强，接近0代表线性无关。")





print("实验1. 行索引是否为姓名：", df.index.name == "Name")
print("实验2. 是否包含Total、Average列：", all(col in df.columns for col in ["Total","Average"]))
print("实验3. Carol记录loc与iloc是否相等：", df.loc["Carol"].equals(df.iloc[2]))
print("实验4. 分组后新增Group_Python_Mean是否存在：", "Group_Python_Mean" in df.columns)
print("实验5. 协方差矩阵形状校验：", cov_matrix.shape == (3,3))
```
