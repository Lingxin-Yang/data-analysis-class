欧⼏⾥得算法

```python

def gcd(m, n):
    while n != 0:
        m, n = n, m % n
    return m


print("GCD of (2018, 2019):", gcd(2018, 2019))
print("GCD of (1200, 300):", gcd(1200, 300))
print("GCD of (5040, 60):", gcd(5040, 60))
```

近似欧拉数e

```python
import math


def euler_limit(n):
    return (1 + 1/n) ** n

# 前n项和
def euler_infinite_sum(n):
    if n == 0:
        return 0.0
    sum_e = 0.0
    for k in range(n): 
        sum_e += 1 / math.factorial(k)
    return sum_e


def euler_approx(epsilon):
    n = 1
    true_e = math.exp(1)
    
    while abs(true_e - euler_infinite_sum(n)) >= epsilon:
        n += 1
    return euler_infinite_sum(n)


def print_euler_sum_table(n):
    for k in range(1, n + 1):
        print(euler_infinite_sum(k))
```

测试整数属性

```python

def is_power_of_2(n):
    
    if n <= 0:
        return False
    while n > 1:
        if n % 2 != 0:
            return False
        n = n // 2     
    return True


def is_power(b, n):
    if n <= 0 or b <= 0:
        return False
    if b == 1:
        return n == 1
    while n > 1:
        if n % b != 0:
            return False
        n = n // b
    return True
```

```python
print(is_power_of_2(8))  
print(is_power_of_2(10)) 
print(is_power(3, 27))   
```
