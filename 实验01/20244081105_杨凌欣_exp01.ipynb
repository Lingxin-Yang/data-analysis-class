实验 2：函数、条件、循环与递归

```python
print("任务1")
def score_level(score):
    if not (0 <= score <= 100):
        raise ValueError("分数超出0~100范围")
    if score >= 90:
        return "优秀"
    elif score >= 80:
        return "良好"
    elif score >= 60:
        return "及格"
    else:
        return "不及格"

# 测试正常和异常情况
print(f"93分的等级是: {score_level(93)}")
print(f"85分的等级是: {score_level(85)}")
print(f"73分的等级是: {score_level(73)}")
print(f"59分的等级是: {score_level(59)}")
try:
    score_level(105)
except ValueError as e:
    print(f"测试105分时捕获异常: {e}")




print("\n任务2")
sum_for = 0
for i in range(1, 101):
    sum_for += i

# 使用 while 循环
sum_while = 0
j = 1
while j <= 100:
    sum_while += j
    j += 1
    
print(f"for 循环结果: {sum_for}")
print(f"while 循环结果: {sum_while}")
print(f"两种方法结果是否一致: {sum_for == sum_while}")





print("\n任务3")
def factorial_recursive(n):
    if n == 0 or n == 1:
        return 1
    return n * factorial_recursive(n - 1)

def factorial_iterative(n):
    res = 1
    for i in range(1, n + 1):
        res *= i
    return res

test_nums = [0, 1, 5, 10]
for n in test_nums:
    print(f"{n}! -> 递归结果: {factorial_recursive(n)}, 迭代结果: {factorial_iterative(n)}")





print("\n任务4")
def summary(values):
    return min(values), max(values), sum(values)/len(values)

test_list = [70, 80, 90, 100, 60]
min_val, max_val, avg_val = summary(test_list)
print(f"列表 {test_list} 的统计结果:")
print(f"最小值: {min_val}, 最大值: {max_val}, 平均值: {avg_val}")






print("\n任务5")
def sum_all(*args):
    return sum(args)

print(f"无参数测试 sum_all(): {sum_all()}")
print(f"单参数测试 sum_all(10): {sum_all(10)}")
print(f"多参数测试 sum_all(1, 2, 3, 4, 5): {sum_all(1, 2, 3, 4, 5)}")
```

实验 4 (文件读写、异常处理与模块)

```python
import os, pickle

print("任务1")
cwd = os.getcwd()
data_dir = os.path.join(cwd, "exp01_data")
os.makedirs(data_dir, exist_ok=True)
#写入数据
file_path = os.path.join(data_dir, "students.txt")
raw_data = """name, score
Alice, 85
Bob, 58
Carol, 92
David, 76
Eva, 60
Frank, 89
Grace, abc
Henry, 105
Incomplete"""


# 任务2
with open(file_path, "w", encoding="utf-8") as f:
    f.write(raw_data)

#任务3
# 解析函数
def parse_student(line):
    parts = [p.strip() for p in line.split(',')]
    if len(parts) != 2:
        raise ValueError("字段数量不正确")
    name, score_str = parts
    if not name:
        raise ValueError("姓名为空")
    score = int(score_str)
    if not (0 <= score <= 100):
        raise ValueError("成绩不在0~100范围内")
    return {"name": name, "score": score}





# 任务4
valid_records = []
invalid_count = 0

with open(file_path, "r", encoding="utf-8") as f:
    lines = f.readlines()
    for line_num, line in enumerate(lines[1:], start=2): 
        line = line.strip()
        if not line: continue
        try:
            valid_records.append(parse_student(line))
        except ValueError as e:
            print(f"行号 {line_num} 错误: {e} -> 内容: '{line}'")
            invalid_count += 1
print(f"有效记录数: {len(valid_records)}")




print("\n任务5")
report_path = os.path.join(data_dir, "report.txt")
avg = sum(r['score'] for r in valid_records) / len(valid_records)
pass_rate = sum(1 for r in valid_records if r['score'] >= 60) / len(valid_records)
stats_str = f"有效: {len(valid_records)}, 无效: {invalid_count}, 平均分: {avg:.2f}, 及格率: {pass_rate:.1%}\n"

with open(report_path, "w", encoding="utf-8") as f:
    f.write("W模式(覆盖之前的所有内容)\n" + stats_str)
with open(report_path, "a", encoding="utf-8") as f:
    f.write("A模式(在已有内容末尾追加)\n" + stats_str)
with open(report_path, "r", encoding="utf-8") as f:
    print("report.txt 当前内容:\n" + f.read().strip())


print("\n任务6")
pkl_path = os.path.join(data_dir, "students.pkl")
with open(pkl_path, "wb") as f:
    pickle.dump(valid_records, f)
with open(pkl_path, "rb") as f:
    loaded_records = pickle.load(f)
print(f"Pickle恢复的数据与原数据相等: {loaded_records == valid_records}")





print("\n任务7")
module_code = """
def parse_student(line):
    parts = [p.strip() for p in line.split(',')]
    if len(parts) != 2: raise ValueError("字段数量不正确")
    name, score = parts[0], int(parts[1])
    if not (0 <= score <= 100): raise ValueError("成绩范围错误")
    return {"name": name, "score": score}
"""
with open("student_utils.py", "w", encoding="utf-8") as f:
    f.write(module_code.strip())

import student_utils
print(f"模块有效测试: {student_utils.parse_student('Alice, 85')}")
try:
    student_utils.parse_student('Error, abc')
except ValueError as e:
    print(f"模块无效测试捕获: {e}")





print("\n任务8")
try:
    open("not_exist_file.txt", "r")
except FileNotFoundError as e:
    print(f"捕获异常: {e}")
```

实验 5 (类、继承与对象复制)

```python
import copy

print("任务1和2")
class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score
        
    def is_passed(self):
        return self.score >= 60
        
    def __str__(self):
        return f"Student(Name: {self.name}, Score: {self.score})"

# 使用上一步的有效记录创建对象
students = [Student(r["name"], r["score"]) for r in valid_records]
for s in students[:3]:  # 打印前3个
    print(f"{s}, 及格状态: {s.is_passed()}")




print("\n任务3")
class GraduateStudent(Student):
    def __init__(self, name, score, research_topic):
        super().__init__(name, score)  # 初始化父类属性
        self.research_topic = research_topic
        
    def __str__(self):
        return f"GraduateStudent(Name: {self.name}, Score: {self.score}, Topic: {self.research_topic})"

grad = GraduateStudent("Tom", 88, "Artificial Intelligence")
print(grad)




print("\n任务4")
# 1. 直接赋值
orig1 = {"name": "Alice", "scores": [85, 90]}
assign_obj = orig1
assign_obj["scores"].append(100)
print(f"直接赋值修改后，原对象变为: {orig1}")

# 2. copy.copy()
orig2 = {"name": "Alice", "scores": [85, 90]}
shallow_copy = copy.copy(orig2)
shallow_copy["scores"].append(100)
print(f"浅拷贝修改内层列表后，原对象变为: {orig2} (列表地址相同，受影响)")

# 3. copy.deepcopy()
orig3 = {"name": "Alice", "scores": [85, 90]}
deep_copy = copy.deepcopy(orig3)
deep_copy["scores"].append(100)
print(f"深拷贝修改内层列表后，原对象变为: {orig3} (不受影响)")
```
