#### 实验一，正则表达式的基础匹配

```python
import re

print("任务1 ")
text1 = "cat on mat; later: cats and dogs"
res_d = re.findall(r"\d+", text1)
res_s = re.findall(r"\s+", text1)
res_bcat = re.findall(r"\bcat\b", text1)
print(f"匹配数字：{res_d}")
print(f"匹配空白：{res_s}")
print(f"匹配独立单词cat：{res_bcat}")



print("\n任务2")
text2 = "cat dog bird CAT"
res_or1 = re.findall(r"cat|dog", text2)
print(f"r'cat|dog' 在 '{text2}'：{res_or1}")

res_a1 = re.findall(r"a|aa|aaa", "aaaa")
res_a2 = re.findall(r"aaa|aa|a", "aaaa")
print(f"r'a|aa|aaa' 匹配 'aaaa'：{res_a1}")
print(f"r'aaa|aa|a' 匹配 'aaaa'：{res_a2}")


print("\n任务3")
text3 = "ct cat caat caaat"
p1 = re.findall(r"ca*t", text3)
p2 = re.findall(r"ca+t", text3)
p3 = re.findall(r"ca?t", text3)
p4 = re.findall(r"ca{2}t", text3)
p5 = re.findall(r"ca{1,2}t", text3)
print(f"ca*t  : {p1}  ")
print(f"ca+t  : {p2}   ")
print(f"ca?t  : {p3}   ")
print(f"ca{{2}}t: {p4} ")
print(f"ca{{1,2}}t: {p5} ")
```

#### 实验2：分组、回溯引⽤与替换

```python
import re

print("任务1")
text1 = "联系：alice@example.com ; 备选：bob@test.org"
pattern1 = r"(\w+)@([\w.]+)"
res1 = re.findall(pattern1, text1)
print(res1)



print("\n任务2")
text2 = "cat cat, dog cat, bird bird"

pattern2 = r"\b(\w+)\s+\1\b"
res2 = re.findall(pattern2, text2)
print(res2)



print("\n任务3")
text3 = "2024-05-01\n2024/05/02\n2024.05.03"

pattern3 = r"(\d{4})[-/.](\d{2})[-/.](\d{2})"

result3 = re.sub(pattern3, r"\2-\3-\1", text3)
print("转换前：")
print(text3)
print("转换后：")
print(result3)


print("\n任务4")
text4 = "联系我：123-456-7890 或 987.654.3210"
pattern4 = r"(\d{3})[-.](\d{3})[-.](\d{4})"
result4 = re.sub(pattern4, r"(\1) \2-\3", text4)
print("转换前：", text4)
print("转换后：", result4)
```

#### 实验3：使⽤BeautifulSoup解析HTML

```python
from bs4 import BeautifulSoup

html_doc = """
<html><head><title>实验课程</title></head><body>
<h1>数据分析</h1>
<p class="course">第六讲：正则表达式</p>
<p class="course">第七讲：结构化数据</p>
<a href="https://example.org/lecture06">讲义六</a>
<a href="https://example.org/lecture07">讲义七</a>
</body></html>
"""

# 任务1
soup = BeautifulSoup(html_doc, "html.parser")
title_tag = soup.find("title")
h1_tag = soup.find("h1")
print("任务1")
print("页面标题文本：", title_tag.get_text())
print("h1标签名：", h1_tag.name)
print("h1文本内容：", h1_tag.get_text())

# 任务2
print("\n任务2")
p_list = soup.find_all("p", class_="course")
print("class='course' 的段落：")
for p in p_list:
    print(p.get_text())

a_list = soup.find_all("a")
print("\n所有链接：")
for a in a_list:
    txt = a.get_text()
    link_href = a.get("href")
    print(f"链接文字：{txt}，href：{link_href}")

# 任务3
print("\n任务3")
print("整个文档get_text():")
print(repr(soup.get_text()))

print("\n第一个course段落get_text():")
print(repr(p_list[0].get_text()))

print("\n第一个链接get_text():")
print(repr(a_list[0].get_text()))
```
